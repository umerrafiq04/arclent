```python
# 5d. Insert seed jobs as published rows — reuses the exact same code path as a real
# publish (upsert_job_draft + finalize_publish), so Job ID allocation/sequence state
# stays consistent with anything published later through the live app. Idempotent:
# skips a seed job if a published job with that title already exists for the company.

from backend.database import upsert_job_draft, finalize_publish

def _company_id(name: str) -> int:
    with get_connection() as conn:
        row = conn.execute("SELECT id FROM company_profile WHERE company_name = ?", (name,)).fetchone()
        return row["id"]

def _already_seeded(company_id: int, job_title: str) -> bool:
    with get_connection() as conn:
        row = conn.execute(
            "SELECT id FROM jobs WHERE company_id = ? AND job_title = ? AND status = 'published'",
            (company_id, job_title),
        ).fetchone()
        return row is not None

seeded_count = 0
for spec in SEED_JOBS:
    company_id = _company_id(spec["company_name"])
    if _already_seeded(company_id, spec["job_title"]):
        continue

    session_id = f"seed-{spec['company_name'].lower().replace(' ', '-')}-{spec['job_title'].lower().replace(' ', '-')}"
    job_state = {
        "job_title": spec["job_title"],
        "job_category": spec["job_category"],
        "experience": spec["experience"],
        "location": spec["location"],
        "work_mode": spec["work_mode"],
        "employment_type": spec["employment_type"],
        "required_skills": spec["required_skills"],
        "preferred_skills": spec["preferred_skills"],
        "responsibilities": spec["responsibilities"],
        "company_overrides": {},
    }
    upsert_job_draft(session_id, company_id, job_state, jd_stale=False)

    selected_jd = {
        "job_title": spec["job_title"],
        "requisition_id": None,
        "job_category": spec["job_category"],
        "employment_type": spec["employment_type"],
        "location": spec["location"],
        "work_mode": spec["work_mode"],
        "company_overview": None,
        "minimum_requirements": [],
        "preferred_qualifications": [],
        "stand_out": [],
        **spec["jd"],
    }
    record = finalize_publish(session_id, company_id, spec["company_name"], selected_jd, "1")
    seeded_count += 1
    print(f"  Seeded {record['job_id']}: {spec['job_title']} @ {spec['company_name']}")

print(f"Seeded {seeded_count} new published job(s) (skipped any already present).")
```
